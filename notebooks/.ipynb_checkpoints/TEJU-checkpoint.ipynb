```python
import pandas as pd
import numpy as np
import sklearn
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

print("All packages imported successfully!")
print("Joblib version:", joblib.__version__)
print("Scikit-learn version:", sklearn.__version__)
```

```python
data_path = "../data/customer_churn.csv"
df = pd.read_csv(data_path)
print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
# Understand the Target Variable
df["Churn"].unique()
```

```python
features = [
    "Age",
    "Tenure",
    "MonthlyBill",
    "SupportCalls"
]
target = "Churn"
X = df[features]  #  return  only given columns values 
Y = df[target]     # return only churn column values 
print( X.head() )
print( Y.head() )
```

```python
display(Y.head())
```

```python
# Split Data -  training data (learning) and test data (evaluate)
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42,
    stratify=Y
)
print("Total records :", len(X))
print("Training data :", len(X_train))
print("Testing data  :", len(X_test))
```

```python
# scale the feature 
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Create the ML Model
model = LogisticRegression(
    random_state=42
)
```

```python
# train the model
model.fit(
    X_train_scaled,
    Y_train
)
print("Model training completed!")
```

```python
# Make Predictions
Y_pred = model.predict(
    X_test_scaled
)
print("Predictions:")
print(Y_pred)
```

```python
# compare actual vs predicted
comparison = pd.DataFrame({
    "Actual": Y_test.values,
    "Predicted": Y_pred
})
comparison
```

```python
import joblib
#Save the Trained Model
model_path = "../models/churn_model.pkl"
joblib.dump(
    model,
    model_path
)
print("Model saved successfully!")
```

```python
# save the scalar
scaler_path = "../models/scaler.pkl"
joblib.dump(
    scaler,
    scaler_path
)
print("Scaler saved successfully!")
```

```python
# load  the model 
loaded_model = joblib.load(
    "../models/churn_model.pkl"
)
loaded_scaler = joblib.load(
    "../models/scaler.pkl"
)
print("Model and scaler loaded successfully!")
```

```python
#  New Customer Prediction
# load the new customer
new_customer = pd.DataFrame([
    {
        "Age": 45,
        "Tenure": 5,
        "MonthlyBill": 3500,
        "SupportCalls": 6
    }
])
new_customer
```

```python
# The model was trained using scaled data. 
# So we must scale the new customer's data using the same saved scaler.
new_customer_scaled = loaded_scaler.transform(
    new_customer
)
prediction = loaded_model.predict(
    new_customer_scaled
)
print("Prediction:", prediction)
```

```python
if prediction[0] == 1:
    print("Customer is likely to LEAVE (CHURN).")
else:
    print("Customer is likely to STAY.")
```
